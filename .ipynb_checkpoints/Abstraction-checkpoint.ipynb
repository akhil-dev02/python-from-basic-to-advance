```python
from abc import ABC, abstractmethod
class mobile(ABC):
    @abstractmethod
    def Ram(self):
        pass
    @abstractmethod
    def Rom(self):
        pass
    def battery(self):
        pass
    # concrete method
    def about_device(self):
        return "This is mobile"
# child class
class Oppo(mobile):
    def __init__(self,model,oppo_ram,oppo_rom,oppo_battery):
        self.model = model
        self.oppo_ram = oppo_ram
        self.oppo_rom = oppo_rom
        self.oppo_battery = oppo_battery
    
    def Ram(self):
        return f"The Oppo {self.model} have {self.oppo_ram}GB RAM"
    
    def Rom(self):
        return f"The Oppo {self.model} have {self.oppo_rom}GB ROM"
    def battery(self):
        return f"The Oppo {self.model} have {self.oppo_batter})MAH"
    def camera(self,front, back):
        return f"The Oppo {self.model} have {front}MP front camera + {back}MP Back camera"
# m = mobile() # can;t instanciate for abstract class
reno16 = Oppo("reno16", 16, 256, 5000)
reno15 = Oppo("reno15", 8, 256, 5500)
print(reno16.about_device())
print(reno16.Ram())
print(reno16.Rom())





    
```
