```python
import math
print(dir(math))
```

```python
print(math.ceil(4.956)) # It returns the upper bound value
print(math.floor(4.25)) # it returns the lower bound value
print(math.factorial(5)) # It returns factorial of a number
print(math.gcd(10,5,2))  # It returns the GCD value
print(math.lcm(45,5))    # It returns the LCM value
print(math.log10(2))     # It returns log value
print(math.log10(4))
print(math.log2(4))
print(math.pi)       # It returns pi value
```

```python
import random
print(random.random()) # This Function returns the value in between 0 and 1
print(random.randint(1,10)) # randint(a,b) This function generates random number between 1 to 10
print(random.choice(['A',1,2,3,4,5,6,7,8,'J','Q','K'])) # it returns the random item from sequence
print(random.choices(['A',1,2,3,4,5,6,7,8,'J','Q','K'],k=3))
arr=['A',1,2,3,4,5,6,7,8,'J','Q','K']
random.shuffle(arr)
print(arr)
```

```python
rolls=0
steps=0
while True:
    input("To roll dice for that click enter button")
    dice_val = random.randint(1,6)
    rolls =rolls+1
    if steps + dice_val <=100:
        steps = steps+dice_val

        
    if steps == 100:
        print(f"Your reached TOP in {rolls} rolls")
        break
    print(f"For {rolls} rolls yoyr ar {steps} position")
        
```
