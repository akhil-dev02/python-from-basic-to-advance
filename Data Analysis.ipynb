```python
# Data Transformation
categorical_columns = ['mainroad',
       'guestroom', 'hotwaterheating', 'airconditioning',
        'prefarea', 'furnishingstatus']
def transfer(x):
    if x == 'yes':
        return 1
    else:
        return 0
df['mainroad'] = df['mainroad'].apply(transfer)
df['guestroom'] = df['guestroom'].apply(transfer)
df['hotwaterheating'] = df['hotwaterheating'].apply(transfer)
df['airconditioning'] = df['airconditioning'].apply(transfer)
df['prefarea'] = df['prefarea'].apply(transfer)
df


    
    
```

```python
def transfer(x):
    if x == 'unfurnished':
        return 0
    elif x == 'semi-furnished':
        return 0.5
    else:
        return 1

df['furnishingstatus'] = df['furnishingstatus'].apply(transfer)


def transfer1(x):
    if x == 'yes':
        return 1
    else:
        return 0

df['basement'] = df['basement'].apply(transfer1)

df
```
