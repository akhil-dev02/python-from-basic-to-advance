```python
from datetime import datetime,timedelta
curr_timestamp = datetime.now() # to get current timestamp
print(curr_timestamp)
today =curr_timestamp.date()# It extracts date from timestamp
curr_time = curr_timestamp.time() # it extract time
print(today)
print(curr_time)
print(today.year)
print(today.month)
print(today.day)
print(curr_time.hour,curr_time.minute,curr_time.second)
print(today.strftime("%d-%m-%Y"))
print(curr_time.strftime("%I:%M %p"))
```

```python
import time
from IPython.display import clear_output
seconds = 0
minutes = 0
hours = 0
while True:
    print(f"{hours}:{minutes}:{seconds}")
    time.sleep(1)
    seconds +=1
    if seconds == 60:
        minutes+=1
        seconds = 0
        if minutes == 60:
            hours+=1
            minutes=0
            if hours == 24:
                hours = 0
    print(f"{hours}:{minutes}:{seconds}")
    clear_output(wait=True)
   
    
    

    
    
```
