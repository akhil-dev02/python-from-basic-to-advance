```python
# generator defination
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res=gen(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
print(next(res))

```

```python
# To overcome the stopiteration error we use for loop
def gen(x):
    x=x+1
    yield x
    x=x+2
    yield x
    x=x+3
    yield x
res = gen(5)
for i in res:
    print(i)
```

```python
# return 1 to 10 num using function
def fun(n):
    res =[]
    for i in range(1, n+1):
        res.append(i)
    return res
print(fun(10))

```

```python
# return 1 to 10 num using generators
def gen(x):
    res=[]
    for i in range(1,x+1):
        yield i
res = gen(10)
for i in res:
    print(i)
    
```
