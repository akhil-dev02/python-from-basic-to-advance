```python
# List Comprehension
# Syntax: [exp for iterator in iterable if condition]
# return 1-n numbers as list using list comprehensionn
n = int(input())
res=[x for x in range(1,n)]
print(res)
```

```python
# Find the first n even number using list comprehension
n=int(input())
res=[2 * x for x in range(n)]
print(res)
```

```python
# Find the first n even number using list comprehension
n=int(input())
res=[x for x in range(0, 2*n, 1) if x%2==0]
print(res)
```

```python
# Find which one even or odd,for list of integers
arr = [20,34,33,55,23,54]
res=["even" if x%2==0 else "odd" for x in arr] 
print(res)
```

```python
# even - add 1
# odd - add 0.5
arr = [20,34,33,55,23,54]
res=[x+1 if x%2==0 else x+0.5 for x in arr] 
print(res)
```

```python
# find one dice combinations
print([x for x in range(1,7)])
```

```python
# Find Two dices combinations
print([(i,j) for i in range(1,7) for j in range(1,7)])
```

```python
# Find two dices combinations if any dice face is 4
print([(i,j) for i in range(1,7) for j in range(1,7) if i==4 or j==4])
```
