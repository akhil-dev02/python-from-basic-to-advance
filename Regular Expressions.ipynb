```python
import re
```

```python
# match(pat, string)
string = "The sky is blue and Ocean is blue"
res = re.match("[A-Z]", string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # It returns matched pattern val
```

```python
# match(pat, string)
string = "The sky is blue and Ocean is blue"
res = re.match("[A-Za-z]+", string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # It returns matched pattern val
```

```python
# match(pat, string)
try:
    string = "The sky is blue and Ocean is blue"
    res = re.match("z", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # It returns matched pattern val
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# search(pat, string)
# search 5 length word
string = "The sky is blue and Ocean is blue"
res = re.search("[A-Za-z]{5}", string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # It returns matched pattern val
```

```python
# get the words from string which length is morethan 3
# findall(pat, string)
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.findall("[A-Za-z]+", string)
    print(res)
except Exception as e:
    print("There is no matched pattern:", e)
    
    
```

```python
# validate mobile number
# All are digits,length of number is 10, starts with 6,7,8,9, Reamining are 0 - 9
import re
try:
    mobile = input()
    res = re.findall("^[6-9][0-9]{9}", mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is Invalid")
except Exception as e:
    print("There is no matched pattern:", e)
    
```

```python
# PAN Number validation
# it allow upper and digits, length is 10, first 5 are upper, next 4 are digits, and last one ends with alphabet
import re
try:
    pan = input()
    res = re.findall("^[A-Z]{5}[0-9]{4}[A-Z]", pan)
    print(res)
    if res:
        print(f"{pan} is valid")
    else:
        print(f"{pan} is Invalid")
except Exception as e:
    print("There is no matched pattern:", e)
    

```

```python
# Email validation
#TO validate the email
try:
    email=input()
    res=re.fullmatch(r"^[a-z0-9][a-z0-9.]+\@[a-z0-9.]+\.[a-z]{2,3}",email)
    print(res)
    if res:
        print(f"{email} is valid")
    else:
        print(f"{email} is invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# Date validation
#TO validate the Date
try:
    date=input()
    res=re.fullmatch(r"^(0?[1-9]|[12][0-9]3[01])[-/](0?[1-9]|1[0-2])[-/]([0-9]){4}",date)
    print(res)
    if res:
        print(f"{date} is valid")
    else:
        print(f"{date} is invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```
