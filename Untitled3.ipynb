```python
seat_type = input()
booking_days = int(input())
festival = input()
age  = int(input())
Base_price = 5000
if seat_type == "Business seat":
    Base_price = Base_price +(Base_price * 0.4)
elif seat_type == "Premium Eonomy":
    Base_price = Base_price +(Base_price * 0.2)
elif seat_type ==  "Eonomy":
    Base_price = Base_price
if booking_days > 30:
    Base_price = Base_price * 0.90
elif booking_days < 7:
    Base_price = Base_price * 1.25
if festival:
    Base_price = Base_price * 1.20
if age > 60:
    Base_price = Base_price * 0.85
print(Base_price)
    

    
    

```
