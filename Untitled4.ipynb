```python
base_premium = 10000

age = int(input())
health_score = int(input())
vehicle_type = input()

# Age factor
if age < 25:
    premium = base_premium * 1.20
elif age > 50:
    premium = base_premium * 1.15
else:
    premium = base_premium

# Health score factor
if health_score >= 80:
    premium = premium * 0.90
elif health_score < 60:
    premium = premium * 1.20

# Vehicle type factor
if vehicle_type == "Sports":
    premium = premium * 1.30
elif vehicle_type == "SUV":
    premium = premium * 1.15

print(premium)

    
    

```
